  <h2 style="margin:0 0 6px 0; color:#2d4a8a;">📋 Project 3 — Phecode-Based Case-Control Cohorts</h2>
  <p>For project 3, you will be required to make Phecode-based case-control cohorts for <b>Type 2 diabetes (T2D)</b>.
  T2D is a common complex disease that affects ~15–25% of the human population.</p>
  <p><b>The Phecode for T2D is <code>250.2</code>.</b></p>
</div>

  <h3 style="margin:0 0 10px 0; color:#2d4a8a;">📦 Deliverables</h3>
  <p>Your final deliverable is expected to be a <b>text entry in Canvas</b>. Every student must submit their own
  text entry, but you can work with your group members on the analysis.</p>
  <p>The entry will look like this:</p>
  <pre style="background:#fff; border:1px solid #c9d6ec; border-radius:5px; padding:10px 14px; margin:0 0 12px 0;">Case 20345
Control 50987
Excluded 28668</pre>
  <p style="margin:0;">You can also submit your R or Python code. If you worked on the code with your team, add a note
  in the comments with your group number. Everyone in the group must still submit their own code/answer.</p>
</div>

<div style="background:#eef3fb; border-left:4px solid #4a72c9;
            padding:18px 22px; margin:12px 0; line-height:1.6; font-family:sans-serif;">
  <h3 style="margin:0 0 10px 0; color:#2d4a8a;">🧭 Directions</h3>
  <p>You will create the cohorts based on the following conditions:</p>
  <ol style="margin:0; padding-left:20px;">
    <li style="margin-bottom:8px;"><b>Rule of double</b> — a person with at least 2 entries of any T2D ICD10CM code, registered on 2 different dates, is a <b>Case</b>. The same code counted twice on different dates still qualifies.</li>
    <li style="margin-bottom:8px;">Individuals with just 1 T2D ICD10CM code are ambiguous and should be <b>Excluded</b>.</li>
    <li style="margin-bottom:8px;">Individuals with no T2D codes, but with any code from the exclusion criteria, should be <b>Excluded</b>.</li>
    <li style="margin-bottom:0;">Individuals with no ICD10CM codes belonging to T2D or related conditions (defined in the exclude range) are <b>Controls</b>.</li>
  </ol>
</div>
